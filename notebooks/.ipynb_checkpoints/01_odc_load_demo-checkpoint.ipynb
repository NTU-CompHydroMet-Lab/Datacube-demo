# ODC Sentinel-2 Land-Cover Load Demo

This notebook loads a fixed Greater Taipei bounding box from the indexed Taiwan Sentinel-2-derived annual land-cover GeoTIFFs.

`x` and `y` are spatial coordinates. The land-cover class codes are stored in the `classification` data variable. NoData is class `0`.

```python
from datacube import Datacube
from rasterio.warp import transform_bounds
import matplotlib.pyplot as plt
from matplotlib.colors import BoundaryNorm, ListedColormap
import matplotlib.patches as mpatches

PRODUCT = "s2_landcover_taiwan"
MEASUREMENT = "classification"
OUTPUT_CRS = "EPSG:32651"
RESOLUTION = (-10, 10)

dc = Datacube(app="01_odc_load_demo")
```

```python
dc.list_products().loc[[PRODUCT]]
```

```python
datasets = list(dc.index.datasets.search(product=PRODUCT))
len(datasets), datasets[:1]
```

## Greater Taipei Bbox

The bbox below is an approximate Greater Taipei area in lon/lat, covering Taipei, New Taipei, and Keelung. It is transformed to EPSG:32651 before loading from ODC.

```python
# Approximate Greater Taipei bbox: west, south, east, north in EPSG:4326
taipei_bbox_lonlat = (121.25, 24.75, 122.05, 25.35)

min_x, min_y, max_x, max_y = transform_bounds(
    "EPSG:4326",
    OUTPUT_CRS,
    *taipei_bbox_lonlat,
    densify_pts=21,
)

taipei_query = {
    "x": (min_x, max_x),
    "y": (min_y, max_y),
}
taipei_query
```

```python
taipei = dc.load(
    product=PRODUCT,
    measurements=[MEASUREMENT],
    x=taipei_query["x"],
    y=taipei_query["y"],
    crs=OUTPUT_CRS,
    output_crs=OUTPUT_CRS,
    resolution=RESOLUTION,
)
taipei
```

```python
classification = taipei[MEASUREMENT]
classification
```

## Area Summary

This uses the loaded Greater Taipei bbox. NoData value `0` is masked before counting pixels.

```python
class_names = {
    1: "Water",
    2: "Trees",
    4: "Flooded Vegetation",
    5: "Crops",
    7: "Built Area",
    8: "Bare Ground",
    9: "Snow/Ice",
    10: "Clouds",
    11: "Rangeland",
}

pixel_area_m2 = 100
valid = classification.where(classification != 0).to_series().dropna().astype("uint8")
counts = valid.groupby(level="time").value_counts().rename("pixel_count")
area = counts.to_frame()
area["area_km2"] = area["pixel_count"] * pixel_area_m2 / 1_000_000
area = area.reset_index().rename(columns={MEASUREMENT: "class_code"})
area["class_name"] = area["class_code"].map(class_names)
area
```

## Plot

The plot uses one time slice and decimates pixels for display only. The original loaded data remains at 10 m resolution.

```python
class_colors = {
    1: "#419BDF",
    2: "#397D49",
    4: "#7A87C6",
    5: "#E49635",
    7: "#C4281B",
    8: "#A59B8F",
    9: "#B39FE1",
    10: "#FFFFFF",
    11: "#E3E2C3",
}

codes = list(class_colors)
bounds = [0.5, 1.5, 2.5, 4.5, 5.5, 7.5, 8.5, 9.5, 10.5, 11.5]
cmap = ListedColormap([class_colors[code] for code in codes])
norm = BoundaryNorm(bounds, cmap.N)

plot_time_index = 0
display_step = 10

plot_data = classification.isel(time=plot_time_index).where(classification.isel(time=plot_time_index) != 0)
plot_data = plot_data.isel(y=slice(None, None, display_step), x=slice(None, None, display_step))

fig, ax = plt.subplots(figsize=(10, 9))
plot_data.plot.imshow(ax=ax, cmap=cmap, norm=norm, add_colorbar=False)
time_value = str(classification.time.values[plot_time_index])[:4]
ax.set_title(f"Greater Taipei Land Cover - {time_value}")
ax.set_xlabel("Easting, EPSG:32651")
ax.set_ylabel("Northing, EPSG:32651")

patches = [
    mpatches.Patch(color=class_colors[code], label=f"{code} {class_names[code]}")
    for code in codes
]
ax.legend(handles=patches, loc="center left", bbox_to_anchor=(1.02, 0.5), frameon=False)
plt.tight_layout()
plt.show()
```
